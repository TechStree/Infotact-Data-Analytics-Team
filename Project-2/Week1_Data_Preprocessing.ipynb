```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
```

```python
df["Date"].dtype
```

```python
df = df.dropna(subset=["Date"])
```

```python
df = df.set_index("Date")
```

```python
df.head()
```

```python
df = df.sort_index()
```

```python
df.head()
```

```python
daily_df = df.groupby(df.index).agg({
    "Inventory Level":"mean",
    "Units Sold":"sum",
    "Units Ordered":"sum",
    "Demand Forecast":"sum",
    "Price":"mean",
    "Discount":"mean",
    "Competitor Pricing":"mean"
})
```

```python
daily_df.head()
```

```python
full_range = pd.date_range(
    start=daily_df.index.min(),
    end=daily_df.index.max(),
    freq="D"
)
```

```python
missing_dates = full_range.difference(daily_df.index)

print("Expected Dates :", len(full_range))
print("Available Dates :", len(daily_df.index))
print("Missing Dates :", len(missing_dates))
```

```python
daily_df = daily_df.reindex(full_range)

daily_df.index.name="Date"
```

```python
daily_df.isnull().sum()
```

```python
daily_df = daily_df.interpolate(
    method="time",
    limit_direction="both"
)
```

```python
daily_df.isnull().sum()
```

```python
daily_df.index.is_monotonic_increasing
```

```python
type(daily_df.index)
```

```python
daily_df.to_csv("cleaned_supply_chain_dataset.csv")
```
